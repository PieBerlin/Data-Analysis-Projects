```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data=pd.read_csv('Spotify Youtube Dataset.csv')
```

```python
data.head()
```

```python
data.info
```

```python
data.columns
```

```python
#drop columns of no use
data.drop(columns=['Unnamed: 0','Url_spotify','Uri','Url_youtube'],inplace=True)
```

```python
data.columns
```

```python
data
```

```python
#check missing value
data.isna().sum()
```

```python
#filling the missing values with 0 in likes and comments
data.Likes.isna().sum()
```

```python
data['Likes']=data.Likes.fillna(0)
```

```python
data['Comments']=data.Comments.fillna(0)
```

```python
data.Likes.isna().sum()
```

```python
#drop records for missing records
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

## Q3.What are the most common Album Types on Spotify?How many tracks belong to each album type

```python
data.head(2)
```

```python
data.Album_type.unique()
```

```python
a_type=data.Album_type.value_counts()
a_type
```

```python
#draw pie chart
plt.pie(a_type,labels=a_type.index,autopct="'%1.1f%%'",startangle=60,colors='myr',shadow=True,explode=(0.05,0.05,0.05),pctdistance=0.75)
plt.show()
```

## Q4.How do the Average Views Likes and comments are compared between different album types

```python
data.head(2)
```

```python
# data.Likes=data['Likes'].fillna(0)
# data.Comments=data['Comments'].fillna(0)
```

```python
df=data.groupby('Album_type')[['Likes','Views','Comments']].mean()
```

```python
type(df)
```

```python
df=df.reset_index()
```

```python
df
```

```python
# melt- unpivo a dataframe 
```

```python
df_melted=pd.melt(df,id_vars='Album_type',var_name="Attribute",value_name="Total")
```

```python
#draw bar plot
plt.figure(figsize=(9,4))
sns.barplot(x='Album_type',y='Total',hue='Attribute',data=df_melted)
```

## Q5. Top 5 Youtube channels based on the views?

```python
c_views=data.groupby('Channel')['Views'].sum().sort_values(ascending=False).head(5)
```

```python
c_views_t5=c_views.head(5)
```
