## How much amount the companies spend on R & D?

```python
print("Companies spending for R & D in $Bn")
RD=df.groupby('Company')['R&D_Spending_USD_Mn'].sum()/1000
RD
```

```python
#Draw a bar plot to show the amount on R & D by the companies
plt.bar(df.Company,df['R&D_Spending_USD_Mn'])
plt.show()
# Draws the graph  but we dont know the values in y axis are in millions or etc
```

```python
#second option
plt.bar(RD.index,RD.values,color=['cyan','black','magenta'])
plt.title("R & D Spending by the companies Companies")
plt.xlabel('Company')
plt.ylabel('Amount in USD_$Bn')
plt.show()
```

## Revenue Earned by the companies

```python
df.head()
```

```python
print("Revenue Earned by companies in $USD billions")
rev=df.groupby('Company')['AI_Revenue_USD_Mn'].sum()/1000
rev
```

```python
#Draw a bar graph to show revenues of the companies
plt.bar(rev.index,rev.values,color=['red','blue','green'],width=0.5)
plt.title("Revenue Earned by Companies")
plt.xlabel("Company")
plt.ylabel("Revenue in $Bn")
plt.show()
```

## Expenditure and Revenue Year by year

```python
df.head()
```

```python
spend=df.groupby('Year')['R&D_Spending_USD_Mn'].sum()
spend
```

```python
#Showing the amount spend on R & D
plt.plot(spend.index,spend.values,color='red')
plt.title("Combined R& D Spending Year by Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_$Mn")
plt.show()
```

```python
revenue=df.groupby('Year')['AI_Revenue_USD_Mn'].sum()
revenue
```

```python
#Showing the revenue earned
plt.plot(revenue.index,revenue.values,color='green')
plt.title("Combined Revenue Earned Year by Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_$Mn")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
#Showing the amount spend on R & D
plt.plot(spend.index,spend.values,color='red')
plt.title("Combined R& D Spending Year by Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_$Mn")

plt.subplot(1,2,2)
plt.plot(revenue.index,revenue.values,color='green')
plt.title("Combined Revenue Earned Year by Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_$Mn")
plt.show()

```

```python
plt.figure(figsize=(8,6))

#Showing the amount spend on R & D
plt.plot(spend.index,spend.values,color='red')
plt.plot(revenue.index,revenue.values,color='green')
plt.title("Combined Expenditure vs Revenue Year by Year")
plt.xlabel("Year")
plt.ylabel("Amount in USD_$Mn")
plt.legend(['Expenditure','Revenue'])
plt.show()
```
