## Q7. How many employees Resigned and Terminated in each department

```python
df.head(2)
```

```python
df.Status.unique()
```

```python
df_resigned=df[df.Status=='Resigned']
df_resigned.head(2)
```

```python
Resigned_Empl=df_resigned.groupby("Department")['Status'].count()
Resigned_Empl
```

```python
Resigned_Empl.plot(x=Resigned_Empl.index,y=Resigned_Empl.values,kind='bar',color='black',legend=True,label="Resigned")
plt.title("Resigned employees per depertment")
plt.ylabel("count")
plt.savefig("Resigned_empl_per_dept.png",bbox_inches='tight')
plt.show()
```

```python
df_terminated=df[df.Status=='Terminated']
df_terminated.head(2)
```

```python
type(df_terminated)
```

```python
terminated_empl=df_terminated.groupby("Department")['Location'].count()
terminated_empl
```

```python
type(terminated_empl)
```

```python
terminated_empl.plot(x=terminated_empl.index,y=terminated_empl.values,kind='bar',color='orange',figsize=(8,4))
plt.title("No of Terminated employees per department")
plt.ylabel("Count")
plt.savefig("Terminated_empl_per_dept.png",bbox_inches='tight')

plt.show()
```

```python
Resigned_Empl.plot(x=Resigned_Empl.index,y=Resigned_Empl.values,kind='bar',color='black',legend=True,label="Resigned")
terminated_empl.plot(x=terminated_empl.index,y=terminated_empl.values,kind='bar',color='orange',figsize=(8,4),legend=True,label='Terminated')
plt.title("Terminated and Resigned employees per Depertment")
plt.ylabel("Count Of Employees")
plt.grid()
plt.savefig("TerminatedAndResignedEmployeesPerDept.png",bbox_inches='tight')
plt.show()
```

## Q15.Identify departments with the highest attrition rate (Resigned %)

```python
df.head(2)
```

```python
dept_count=df.groupby('Department')['Status'].agg(total_empl='count',resigned=lambda x:(x=='Resigned').sum())
dept_count
```

```python
type(dept_count)
```

```python
### attrition rate
dept_count['attrition_rate_%']=dept_count.resigned/dept_count.total_empl*100
```

```python
dept_count.sort_values('attrition_rate_%',ascending=False)
```
